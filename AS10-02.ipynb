```python
# Step 1: Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set chart style
sns.set_theme(style="whitegrid")

# Load Sales Dataset
df = pd.read_csv("AS10-02.csv", encoding="latin1")

# Display first 5 rows
print("Sales Dataset Loaded Successfully!")
display(df.head())

# Check dataset dimensions
print("\nDataset Shape:", df.shape)
```

```python
# Step 2: Exploratory Data Analysis (EDA)

# Display column names
print("Column Names:")
print(df.columns.tolist())

# Dataset information
print("\nDataset Information:")
df.info()

# Statistical summary
print("\nStatistical Summary:")
display(df.describe())

# Check number of rows and columns
print("\nDataset Shape:", df.shape)

# Check unique values in important columns
for col in ["Category", "Sub-Category", "Region", "Segment"]:
    if col in df.columns:
        print(f"\nUnique values in {col}:")
        print(df[col].unique())
```

```python
# Step 3: Check Missing Values and Duplicates

# Check missing values
print("Missing Values in Each Column:")
print(df.isnull().sum())

# Total missing values
print("\nTotal Missing Values:")
print(df.isnull().sum().sum())

# Missing values percentage
print("\nMissing Values Percentage:")
print((df.isnull().sum() / len(df) * 100).round(2))

# Check duplicate records
print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
# Step 4: Data Cleaning

# Create a copy of the original dataset
df_clean = df.copy()

# Remove duplicate rows
df_clean = df_clean.drop_duplicates()

# Handle missing values in categorical columns
categorical_cols = df_clean.select_dtypes(
    include=["object"]
).columns

for col in categorical_cols:
    df_clean[col] = df_clean[col].fillna("Unknown")

# Convert date columns if available
for col in ["Order Date", "Ship Date"]:
    if col in df_clean.columns:
        df_clean[col] = pd.to_datetime(
            df_clean[col], errors="coerce"
        )

# Check missing values after cleaning
print("Missing values after cleaning:")
print(df_clean.isnull().sum())

print("\nDuplicate rows after cleaning:")
print(df_clean.duplicated().sum())

print("\nCleaned dataset shape:", df_clean.shape)

display(df_clean.head())
```

```python
# Step 5: Monthly Sales Analysis

# Extract month from Order Date
df_clean["Month"] = df_clean["Order Date"].dt.to_period("M")

# Calculate monthly sales
monthly_sales = (
    df_clean.groupby("Month")["Sales"]
    .sum()
    .reset_index()
)

# Convert Month to string for plotting
monthly_sales["Month"] = monthly_sales["Month"].astype(str)

# Display monthly sales
print("Monthly Sales:")
display(monthly_sales.head(12))

# Create line chart
plt.figure(figsize=(14, 6))

sns.lineplot(
    data=monthly_sales,
    x="Month",
    y="Sales",
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# Step 6A: Profit by Category

category_profit = (
    df_clean.groupby("Category")["Profit"]
    .sum()
    .reset_index()
    .sort_values("Profit", ascending=False)
)

print("Profit by Category:")
display(category_profit)

# Create bar chart
plt.figure(figsize=(10, 5))

sns.barplot(
    data=category_profit,
    x="Category",
    y="Profit"
)

plt.title("Total Profit by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.tight_layout()
plt.show()
```

```python
# Step 6B: Top 10 Products by Profit

product_profit = (
    df_clean.groupby("Sub-Category")["Profit"]
    .sum()
    .reset_index()
    .sort_values("Profit", ascending=False)
    .head(10)
)

print("Top 10 Sub-Categories by Profit:")
display(product_profit)

# Create bar chart
plt.figure(figsize=(12, 6))

sns.barplot(
    data=product_profit,
    x="Profit",
    y="Sub-Category"
)

plt.title("Top 10 Product Sub-Categories by Profit")
plt.xlabel("Total Profit")
plt.ylabel("Sub-Category")
plt.tight_layout()
plt.show()
```

```python
# Step 7: Customer Segmentation Analysis

# Calculate sales for each customer segment
segment_sales = (
    df_clean.groupby("Segment")["Sales"]
    .sum()
    .reset_index()
    .sort_values("Sales", ascending=False)
)

print("Sales by Customer Segment:")
display(segment_sales)

# Create bar chart
plt.figure(figsize=(9, 5))

sns.barplot(
    data=segment_sales,
    x="Segment",
    y="Sales"
)

plt.title("Total Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
# Create pie chart
plt.figure(figsize=(7, 7))

plt.pie(
    segment_sales["Sales"],
    labels=segment_sales["Segment"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Sales Contribution by Customer Segment")
plt.tight_layout()
plt.show()
```

```python
# Regional Sales Analysis
region_sales = (
    df_clean.groupby("Region")["Sales"]
    .sum()
    .reset_index()
    .sort_values("Sales", ascending=False)
)

print("Sales by Region:")
display(region_sales)

plt.figure(figsize=(9, 5))
sns.barplot(data=region_sales, x="Region", y="Sales")
plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```
